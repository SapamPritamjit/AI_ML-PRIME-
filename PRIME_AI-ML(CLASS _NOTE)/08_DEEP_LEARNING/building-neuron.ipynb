```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42) # act like as a random_state
input = torch.tensor([1.0, 2.0, 3.0])
```

```python
neuron = nn.Linear(in_features = 3, out_features = 1)
```

```python
output = neuron(input)
```
