```python
df.head()
```

```python
# loudness vs energy
sns.scatterplot(x=df["loudness"], y=df["energy"])
```

```python
# loudness vs danceability
sns.scatterplot(x=df["loudness"], y=df["danceability"])
```

```python
df["acousticness"].corr(df["explicit"])
```

```python
# # Acousticness tells you how likely a song is to use acoustic sounds rather than electronic sounds.
# Think:
# Acoustic guitar/piano → high acousticness
# Electronic/EDM/synth-heavy music → low acousticness

sns.scatterplot(x=df["acousticness"], y=df["energy"])
```

```python
df["acousticness"].corr(df["energy"])
```

```python
# Instrumentalness tells you how likely a song is to contain no vocals/lyrics.

sns.scatterplot(x=df["instrumentalness"], y=df["popularity"])
```

```python
df["danceability"].corr(df["popularity"])
```

```python
# Liveness tells you whether the recording sounds like it was performed in front of a live audience.

sns.scatterplot(x=df["liveness"], y=df["danceability"])
```

```python
df["danceability"].corr(df["liveness"])
```

```python
df["mode"].value_counts()
```

```python
# tempo – The speed of the song measured in beats per minute (BPM)

sns.scatterplot(x=df["loudness"], y=df["tempo"])
```

```python
df["loudness"].corr(df["tempo"])
```
